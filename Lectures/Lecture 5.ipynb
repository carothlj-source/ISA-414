```python
def RatioMaxMin(value_1, value_2):
    ratio = max(value_1, value_2) / min(value_1, value_2)
    return ratio
```

#### Slide 10

```python
print(RatioMaxMin(10,2))
print(RatioMaxMin(2, 10))
print(RatioMaxMin(1,2))
```

```python
# you will get an error
print(RatioMaxMin(10,0))
```

```python
# you will get an error
print(RatioMaxMin(10,'0'))
```

```python
def RatioMaxMin2(value_1, value_2 = 10):
    ratio = max(value_1, value_2) / min(value_1, value_2)
    return ratio
```

#### Slide 12

```python
print(RatioMaxMin2(10))
print(RatioMaxMin2(5))
# 2 replaces 10
print(RatioMaxMin2(5,2))
```

#### Slide 13

```python
def RatioMaxMin3(value_1, value_2 = 10):
    if type(value_1) == int and type(value_2) == int:
        ratio = max(value_1, value_2) / min(value_1, value_2)
        return ratio
    else:
        return "One of the arguments is not of type integer"
```

```python
# testing the above function
print(RatioMaxMin3(10,2))
print(RatioMaxMin3(10,"2"))
```

#### Slide 14

```python
def RatioMaxMin3(value_1, value_2 = 10):
    try:
        ratio = max(value_1, value_2) / min(value_1, value_2)
        return ratio
    except:
        return "An error occurred"
```

```python
# testing the above function
print(RatioMaxMin3(10,0))
print(RatioMaxMin3(10,"2"))
print(RatioMaxMin3(10.0,3.6))
```

#### Slide 27

```python
from folium import Map, Marker 
import pandas
```

#### Slide 28

```python
# reading the data
df = pandas.read_csv("web_log.csv")

df
```

```python
# removing rows that contain duplicate combinations of latitude and longitude
df = df.drop_duplicates(["Latitude","Longitude"])
df
```

#### Slide 29

```python
# Create a map centered at a specific location (adjust as needed)
m = Map(zoom_start=1)

# Add markers with names to the map
for index in range(len(df)):
    latitude = df.iloc[index,3]
    longitude = df.iloc[index,4]
    name = df.iloc[index,0]
    marker = Marker([latitude, longitude], popup=name)
    marker.add_to(m)

# Display the map
m
```

Saving the map to HTML

```python
m.save("web_map.html")
```
