#### Slide 8

```python
# first code
# a range of elements to be retrieved
# slicing notation: [start:end:step]
mixed_list[0:6:2]
```

```python
# second code
# by leaving out the start (or end) value, the range will start (or end) at the first (or last) item
# this code starts at the second item
mixed_list[2:]
```

```python
# third code
# this code stop sat the 5th item
mixed_list[:5]
```

#### Slide 11

```python
# first code
# looping over a list of items
for value in combined_list:
    print(value)
```

```python
# second code
# looping through index numbers
for index in range(len(combined_list)):
    print(combined_list[index])
```

#### Slide 12

```python
# create loops to add elements to a list
string_list = []
for value in combined_list:
    if type(value) == str:
        string_list.append(value)

string_list
```

#### Slide 13

```python
# first code
# list comprehension offers a shorter and more efficient way of creating a new list
string_list = [value for value in combined_list if type(value) == str]
string_list
```

```python
# second code
# note that the syntax changes if there is an else statement
     # printing 'none' for values that aren't str
another_list = [value if type(value) == str else None for value in combined_list]
another_list
```
