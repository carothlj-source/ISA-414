```python
string_list = []
for value in combined_list:
    if type(value) == str:
        string_list.append(value)

string_list
```

#### Slide 13

```python
# first code
string_list = [value for value in combined_list if type(value) == str]
string_list
```

```python
# second code
another_list = [value if type(value) == str else None for value in combined_list]
another_list
```
