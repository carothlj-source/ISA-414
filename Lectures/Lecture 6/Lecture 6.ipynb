Note that we are simply reading lines here. This file is *not* a CSV (tabular) file. Thus, there is no need to use **pandas** and read the file as a data frame using the `pandas.read_csv` function.

```python
file = open('access.log.txt', 'r')
web_logs = file.readlines()
```

#### Slide 15

```python
IP = []
for log in web_logs:
    words = log.split(" ")
    IP.append(words[0])

# checking the first six elements of IP
IP[:5]
```

#### Slide 18
Keep in mind that the code below is for only one string (`web_logs[0]`)

```python
import re

# first parameter: pattern
# second parameter: string
hits = re.search(" ", web_logs[0])

# start: initial position where the found pattern starts
pos_first_space = hits.start()
pos_first_space
```

#### Slide 19

Keep in mind that the code below is for only one string (`web_logs[0]`)

```python
hits = re.finditer(" ", web_logs[0])

all_space_positions = []
for hit in hits:
    all_space_positions.append(hit.start())

print(all_space_positions)
```

#### Slide 20

```python
string = web_logs[0]

first_ip = string[0:pos_first_space:1] #you can also write the range as :pos_first_space:
first_ip
```

#### Slide 21

```python
IP = []

for log in web_logs:
    hits = re.search(" ", log)
    pos_first_space = hits.start()
    IP.append(log[0:pos_first_space:1])
     
# printing first 6 IPs
print(IP[:6])

# printing last 6 IPs
print(IP[-1:-7:-1])
```
