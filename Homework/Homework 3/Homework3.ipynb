```python
# Load the CSV file generated in Part 1 using Python
import pandas 
randomdata = pandas.read_csv("carothers_lindsey.csv")
randomdata
```

```python
# Create a subset of the data by selecting only the rows whose indices are odd numbers 
odddata = randomdata.iloc[1::2,:]

# Keeping only the column that contains the city names
cityodddata = odddata["city"]
```

```python
# Generate a bar chart that visualizes the number of web requests per city for the top 10 most popular cities
top_10 = cityodddata.value_counts().head(10)
top_10.plot(kind="bar", xlabel="City", ylabel="Requests", legend=False, title="Top 10 Most Popular Cities by Web Requests")
```

```python
# Upload the resulting data set (CSV format)
top_10.to_csv("carothers_lindsey.csv")
```
